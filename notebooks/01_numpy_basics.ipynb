# NumPy Basics

This notebook contains my NumPy practice.

```python
import numpy as np
```

## Creating a Numpy Array

```python
A = np.array([1, 2, 3])
print(A)
 
B = np.array([[1, 2, 3], [4, 5, 6]])
print(B)
 
C = np.array((1, 3, 2))
print(C)
```

## Accessing the Array Index

```python
A = np.array([[-1, 2, 0, 4],[4, -0.5, 6, 0],[2.6, 0, 7, 8],[3, -7, 4, 2.0]])

B = A[:2, ::2]
print("B:",B)

C = A[[1, 1, 0, 3], [3, 2, 1, 0]]
print("C:",C)
```

## Basic Array Operations

```python
a = np.array([[1, 2],[3, 4]])

b = np.array([[4, 3],[2, 1]])

print("a+1: ", a + 1)
print("b-2: ", b - 2)
print("Sum of all elements in a: ", a.sum())
print("a+b: ", a + b)
```

## Data Types in Numpy

```python
x = np.array([1, 2])  
print(x.dtype)         
 
x = np.array([1.0, 2.0]) 
print(x.dtype)  
 
x = np.array([1, 2], dtype = np.int64)   
print(x.dtype)
```

## Math Operations on DataType Array

```python
A = np.array([[4, 7], [2, 6]], dtype=np.float64)
B = np.array([[3, 6], [2, 8]], dtype=np.float64)

print(np.add(A, B))
print(np.sum(A))
print(np.sqrt(A))
print(A.T)
```

## Array Shapes and Dimensions

```python
A = np.array([1, 2, 3])

B = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

C = np.array([
    [1],
    [2],
    [3]
])

print("A shape:", A.shape)
print("A dimensions:", A.ndim)

print("B shape:", B.shape)
print("B dimensions:", B.ndim)

print("C shape:", C.shape)
print("C dimensions:", C.ndim)
```

## Reshaping Arrays

```python
A = np.array([1, 2, 3, 4, 5, 6])

print(A.shape)

B = A.reshape(2, 3)
print(B)
print(B.shape)

C = A.reshape(3, 2)
print(C)
print(C.shape)

D = A.reshape(6, 1)
print(D)
print(D.shape)
```

## Matrix Multiplication

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Element-wise multiplication:")
print(A * B)

print("Matrix multiplication:")
print(A @ B)
```

```python
A = np.random.randn(3, 2)
B = np.random.randn(2, 4)

print("A:", A.shape)
print("B:", B.shape)

C = A @ B

print("C:", C.shape)
```

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print(A * B)
print(A / B)
print(A ** 2)
```

## Sum and Mean Along Axes

```python
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("Total:", np.sum(A))

print("axis=0:", np.sum(A, axis=0))
print("axis=1:", np.sum(A, axis=1))

print("Mean:", np.mean(A))
```

## Broadcasting

```python
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

b = np.array([
    [10],
    [20]
])

print("A shape:", A.shape)
print("b shape:", b.shape)

print(A + b)
```

## Random Numbers and Initialization

```python
A = np.random.randn(3, 2)

print(A)
print(A.shape)
```

```python
np.random.seed(42)

A = np.random.randn(3, 2)

print(A)
```
